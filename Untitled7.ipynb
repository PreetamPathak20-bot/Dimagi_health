```python
df = df.drop_duplicates()
df['Physical_Activity_Hours'] = df['Physical_Activity_Hours'].clip(lower=0)
```

```python
top_countries = df['Country'].value_counts().index[:10].tolist()
```

```python
df.head()
```

```python
def group_countries(country):
 if country in top_countries:
  return country
 else:
  return 'Other'
```

```python
df['Grouped_country'] = df['Country'].apply(group_countries)
```

```python

df['Grouped_country'].value_counts()
```

```python
from sklearn.model_selection import train_test_split
skwewd_col         = ['Study_Hours']
other_numeric_cols = ["Age", "Avg_Daily_Usage_Hours", "Daily_Unlocks", "Physical_Activity_Hours",
                      "Sleep_Hours_Per_Night"]
ordinal_col        = ['Stress_Level']
normal_col         = ["Gender", "Academic_Level", "Most_Used_Platform", "Purpose_Of_Use", 'Grouped_country']

feature_col = skwewd_col + other_numeric_cols + ordinal_col + normal_col
X = df[feature_col]
y = df['Mental_Health_Score']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.30, random_state=42)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import FunctionTransformer,StandardScaler,OrdinalEncoder,OneHotEncoder
from sklearn.compose import ColumnTransformer
#1. Skewed features
skew_pipeline = Pipeline(steps=[
    ('log_transformer', FunctionTransformer(np.log1p)),
    ('scaler', StandardScaler())
])

#2. Numeric Features
plain_numeric_pipeline =Pipeline(steps=[
    ('scaler', StandardScaler())
])

#3. Ordinal
ordinal_pipeline = Pipeline(steps=[
    ('encode', OrdinalEncoder(categories=[['Low','Medium','High','Very High']]))
])

#4. Nominal Features
nominal_pipeline = Pipeline(steps=[
    ('encode', OneHotEncoder(handle_unknown='ignore'))
])

#5 columnTransfromer(konsi_pipeline , konsa feture yani column jayega)
preprocessor = ColumnTransformer = ColumnTransformer(transformers=[
    ('skew_pipeline', skew_pipeline, skwewd_col),
    ('plain_numeric_popeline', plain_numeric_pipeline, other_numeric_cols),
    ('ordinal_pipeline', ordinal_pipeline, ordinal_col),
    ('nominal_pipeline', nominal_pipeline, normal_col)
])
```

```python
print("parikshit")
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, r2_score,mean_squared_error
lr_pipeline = Pipeline(steps=[
('preprocessor', preprocessor),
('model' ,LinearRegression())
])

lr_pipeline.fit(X_train,y_train)
lr_preds = lr_pipeline.predict(X_test)
lr_preds_train = lr_pipeline.predict(X_train)

lr_r2_testing = r2_score(y_test,lr_preds)
lr_r2_trainng = r2_score(y_train,lr_preds_train)
print(f"Accuracy of trainng data is {lr_r2_trainng*100}")
print(f"Accuracy of testing data is {lr_r2_testing*100}")
```

Random forest(rf)

```python
from sklearn.ensemble import RandomForestRegressor
rf_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('model', RandomForestRegressor(random_state=42))
])

rf_pipeline.fit(X_train,y_train)
rf_preds = rf_pipeline.predict(X_test)
rf_preds_train = rf_pipeline.predict(X_train)

rf_r2_testing = r2_score(y_test,rf_preds)
rf_r2_trainng = r2_score(y_train,rf_preds_train)
print(f"Accuracy of trainng data is {rf_r2_trainng*100}")
print(f"Accuracy of testing data is {rf_r2_testing*100}")
```

**Hyperparameter tunning:**

1.   what is best value of parrameter that effect best accuracy of model.



```python
from sklearn.model_selection import RandomizedSearchCV

# FIXED: Replaced single underscores with double underscores (__) between the step name and parameter
param_grid = {
    'model__n_estimators':[100,200,300],
    'model__max_depth':[5,10,15],
    'model__min_samples_split': [2, 5, 10],
    'model__min_samples_leaf': [1, 2, 4]
}

random_search = RandomizedSearchCV(
    estimator=rf_pipeline,
    param_distributions=param_grid,
    n_iter=15,
    cv=5,
    scoring='r2',
    random_state=42,
    n_jobs=-1  # processor free
)

random_search.fit(X_train, y_train)
```

```python
(random_search.best_params_)#this best parameter in which our model outstanding perfomance
```

```python
rf_best_pipeline = random_search.best_estimator_
rf_best_preds   = rf_best_pipeline.predict(X_test)
print(f"R2 Score for random forest after hyperparameter tuning {r2_score(y_test, rf_best_preds)}")
print(f"MAE Score for random forest after hyperparameter tuning {mean_absolute_error(y_test, rf_best_preds)}")

```
